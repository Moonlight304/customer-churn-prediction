```python
# Robust merge of already-engineered feature tables from data/03_final
KEY_COL = 'co_ref'

# Standardize column names up front
for df in [billings, cc_calls, emails, renewals]:
    df.columns = df.columns.str.lower().str.strip()

def _clean_key(df, key_col=KEY_COL):
    out = df.copy()
    out[key_col] = out[key_col].astype('string').str.strip()
    return out

def _dedupe_feature_table(df, source_name, key_col=KEY_COL):
    if key_col not in df.columns:
        raise ValueError(f"{source_name} is missing required key column '{key_col}'")

    work = _clean_key(df, key_col=key_col)
    before_rows = len(work)

    # Remove exact duplicate rows first
    exact_dupes = int(work.duplicated().sum())
    if exact_dupes:
        work = work.drop_duplicates()

    # Pick one row per customer using richest row (most non-null values).
    non_key_cols = [c for c in work.columns if c != key_col]
    work['_non_null_count'] = work[non_key_cols].notna().sum(axis=1)

    sort_cols = ['_non_null_count']
    ascending = [False]

    # If cutoff_date exists, prefer the latest one as tie-breaker.
    if 'cutoff_date' in work.columns:
        work['_cutoff_sort'] = pd.to_datetime(work['cutoff_date'], errors='coerce')
        sort_cols.append('_cutoff_sort')
        ascending.append(False)

    work = (
        work.sort_values(sort_cols, ascending=ascending)
            .drop_duplicates(subset=[key_col], keep='first')
            .drop(columns=['_non_null_count', '_cutoff_sort'], errors='ignore')
    )

    unique_keys = int(work[key_col].nunique(dropna=True))
    duplicate_keys_after = int(work[key_col].duplicated().sum())
    if duplicate_keys_after != 0:
        raise RuntimeError(f"{source_name} still has duplicate {key_col} after dedupe")

    stats = {
        'before_rows': before_rows,
        'after_rows': len(work),
        'exact_dupes_removed': exact_dupes,
        'unique_keys': unique_keys,
    }
    return work, stats

# Prepare base (billings)
merged_df = _clean_key(billings)
base_rows = len(merged_df)
base_unique = int(merged_df[KEY_COL].nunique(dropna=True))
base_missing_key = int(merged_df[KEY_COL].isna().sum() + (merged_df[KEY_COL] == '').sum())

print('billings rows (base):', base_rows)
print('billings unique co_ref:', base_unique)
if base_missing_key:
    print(f"Warning: billings has {base_missing_key} rows with missing/blank {KEY_COL}.")

source_tables = {
    'cc_calls': cc_calls,
    'emails': emails,
    'renewals': renewals,
}

for source_name, source_df in source_tables.items():
    prepared, stats = _dedupe_feature_table(source_df, source_name, key_col=KEY_COL)

    # Indicator: whether this billing row has a match in the source feature table
    matched_mask = merged_df[KEY_COL].isin(prepared[KEY_COL])
    merged_df[f'has_{source_name}_data'] = matched_mask.astype('int8')

    pre_merge_cols = set(merged_df.columns)
    merged_df = merged_df.merge(
        prepared,
        on=KEY_COL,
        how='left',
        suffixes=('', f'_{source_name}'),
        validate='many_to_one'
    )

    if len(merged_df) != base_rows:
        raise RuntimeError(
            f"Row count changed after merging {source_name}: {len(merged_df)} vs {base_rows}"
        )

    added_cols = [c for c in merged_df.columns if c not in pre_merge_cols]
    feature_cols_from_source = [c for c in added_cols if c != f'has_{source_name}_data']

    if feature_cols_from_source:
        all_null_rows = int(merged_df[feature_cols_from_source].isna().all(axis=1).sum())
        all_null_ratio = round(all_null_rows / base_rows, 4)
    else:
        all_null_rows = base_rows
        all_null_ratio = 1.0

    print(
        f"{source_name}: before={stats['before_rows']}, after_dedupe={stats['after_rows']}, "
        f"unique_co_ref={stats['unique_keys']}, matched_rows={int(matched_mask.sum())}, "
        f"unmatched_rows={int((~matched_mask).sum())}, all_source_features_null_rows={all_null_rows} "
        f"({all_null_ratio})"
    )

# Basic sanity checks for recency features
for col in ['days_since_last_payment', 'days_since_last_call', 'days_since_last_cc_call']:
    if col in merged_df.columns:
        neg_count = int((pd.to_numeric(merged_df[col], errors='coerce') < 0).sum())
        if neg_count:
            print(f"Warning: {neg_count} rows have negative values in {col}.")

print('merged rows         :', len(merged_df))
print('merged unique co_ref:', merged_df[KEY_COL].nunique(dropna=True))

merged_df.head()
```
